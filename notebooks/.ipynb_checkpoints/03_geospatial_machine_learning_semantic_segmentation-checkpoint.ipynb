# Geospatial Machine Learning: Semantic Segmentation

This notebook demonstrates how to train a **semantic segmentation** model for pixel-level classification (e.g., building vs. background) in satellite or aerial imagery. We use the [GeoAI](https://opengeoai.org/) library and U-Net with a ResNet34 encoder, following the workflow from the [GeoAI semantic segmentation example](https://opengeoai.org/examples/train_segmentation_model/).

**Note**: Training a neural network is computationally expensive. Expect several minutes to hours depending on your hardware. You can reduce `num_epochs` for a quicker (but less accurate) run.

## What Problem Does Semantic Segmentation Solve?

In geospatial ML, we often need pixel-level classification—labeling each pixel as belonging to a class (e.g., building, water, vegetation).

- **Semantic segmentation**: Labels every pixel with a class (e.g., "building" vs "not building"). All buildings share the same class label; the model does not distinguish between individual objects. Output is a single mask with one value per pixel.
- **Instance segmentation**: Identifies each object separately (Building A, Building B, etc.). Use when you need to count objects or measure each one individually.

| Use Semantic Segmentation when… | Use Instance Segmentation when… |
|--------------------------------|----------------------------------|
| You only need area coverage or pixel-level class maps | You need to count or measure individual objects |
| Features are continuous (e.g., vegetation, water) | Objects may overlap or touch |
| You want faster training and inference | You need spatial relationships between objects |
| Examples: land cover, flood extent, building footprint | Examples: building count, car detection, solar panels |

**Model outputs**: Semantic segmentation gives one class per pixel (binary or multi-class mask). Instance segmentation gives bounding boxes, confidence scores, and a separate mask per object.

## Architecture

**U-Net** (and similar architectures like DeepLabV3, FPN) treats the task as pixel-level classification. It is typically faster to train and run than instance segmentation models like Mask R-CNN.

**U-Net architecture (simple version)**

Follow the diagram below from top to bottom. U-Net looks like the letter "U" because it goes down one side and back up the other.

- **Encoder (Contracting Path)** — Start at **Input Image** at the top. The image flows down through **Conv 64** → **Conv 128** → **Conv 256** → **Conv 512**. Each step makes the image smaller (like zooming out) and learns more about what's in the picture.

- **Bottleneck 1024** — At the bottom, the image is at its smallest. This is where the network understands the whole scene.

- **Decoder (Expanding Path)** — From the bottleneck, the image flows up through **UpConv 512** → **UpConv 256** → **UpConv 128** → **UpConv 64**. Each step makes the image bigger again (like zooming back in) and figures out what label each pixel should get.

- **Skip connections (dotted lines)** — See the dotted arrows labeled "skip"? Those are shortcuts! Conv 64 sends a copy to UpConv 512, Conv 128 to UpConv 256, and so on. The right side uses these copies to remember the details (edges, shapes) that the left side saw before things got shrunk.

- **Output Mask** — The final result: a map where every pixel is labeled (e.g., "building" or "not building").

```python
import mermaid as md

md.Mermaid("""
flowchart TB
    subgraph encoder [Encoder - Contracting Path]
        direction TB
        A[Input Image] --> B[Conv 64]
        B --> C[Conv 128]
        C --> D[Conv 256]
        D --> E[Conv 512]
    end

    E --> F[Bottleneck 1024]

    subgraph decoder [Decoder - Expanding Path]
        direction TB
        F --> G[UpConv 512]
        G --> H[UpConv 256]
        H --> I[UpConv 128]
        I --> J[UpConv 64]
        J --> K[Output Mask]
    end

    B -.->|skip| G
    C -.->|skip| H
    D -.->|skip| I
    E -.->|skip| J
""")
```

```python
# Optional: Install GeoAI and dependencies
# %pip install geoai-py
import geoai
from pathlib import Path
```

## Download Sample Data

We use NAIP (National Agriculture Imagery Program) aerial imagery and building footprint labels from a public Hugging Face dataset. The data includes:
- **Training raster**: RGB aerial image for training
- **Training vectors**: Building polygons (GeoJSON)
- **Test raster**: A separate image for inference

```python
train_raster_url = (
    "https://huggingface.co/datasets/giswqs/geospatial/resolve/main/naip_rgb_train.tif"
)
train_vector_url = "https://huggingface.co/datasets/giswqs/geospatial/resolve/main/naip_train_buildings.geojson"
test_raster_url = (
    "https://huggingface.co/datasets/giswqs/geospatial/resolve/main/naip_test.tif"
)

data_dir = Path("../data")
data_dir.mkdir(exist_ok=True)

train_raster_path = geoai.download_file(train_raster_url, output_path=f"{data_dir}/naip_rgb_train.tif")
train_vector_path = geoai.download_file(train_vector_url, output_path=f"{data_dir}/naip_train_buildings.geojson")
test_raster_path = geoai.download_file(test_raster_url, output_path=f"{data_dir}/naip_test.tif")

print("Downloaded:", train_raster_path, train_vector_path, test_raster_path)
```

## Visualize Sample Data

Inspect the training raster metadata and optionally view the imagery and labels interactively.

```python
geoai.get_raster_info(train_raster_path)
```

```python
# Optional: Interactive map of training data with building outlines
style_dict = {"color": "#ff0000", "weight": 2, "opacity": 1, "fillOpacity": 0}
geoai.view_vector_interactive(
    train_vector_path, tiles=train_raster_path,
    style_function=lambda x: style_dict
)
```

## Create Training Data

The model needs image tiles and corresponding label masks. `export_geotiff_tiles` cuts the raster into fixed-size tiles (512×512) and rasterizes the vector labels into matching mask images. The `stride` controls overlap between tiles.

```python
out_folder = str(data_dir / "buildings_semantic")
tiles = geoai.export_geotiff_tiles(
    in_raster=train_raster_path,
    out_folder=out_folder,
    in_class_data=train_vector_path,
    tile_size=512,
    stride=256,
    buffer_radius=0,
)
print(f"Created {len(tiles)} tiles in {out_folder}")
```

## Train the Semantic Segmentation Model

We train a **U-Net** model with a ResNet34 encoder. Key parameters:
- `architecture="unet"`, `encoder_name="resnet34"`: U-Net is efficient for semantic segmentation
- `num_classes=2`: background + building
- `batch_size=8`: Typically larger than instance segmentation (simpler model)
- `num_epochs=10`: Reduce for faster runs, increase for better accuracy
- `val_split=0.2`: 20% of tiles held out for validation

**Training can take 5–30+ minutes** depending on hardware.

```python
geoai.train_segmentation_model(
    images_dir=f"{out_folder}/images",
    labels_dir=f"{out_folder}/labels",
    output_dir=f"{out_folder}/unet_models",
    architecture="unet",
    encoder_name="resnet34",
    encoder_weights="imagenet",
    num_channels=3,
    num_classes=2,  # background + building
    batch_size=8,
    num_epochs=10,
    learning_rate=0.001,
    val_split=0.2,
    verbose=True,
)
```

## Run Inference

Apply the trained model to the test image. `semantic_segmentation` uses a sliding window to process large rasters. Each pixel is assigned a class label (0 = background, 1 = building).

```python
masks_path = str(data_dir / "naip_test_semantic_prediction.tif")
model_path = f"{out_folder}/unet_models/best_model.pth"

geoai.semantic_segmentation(
    input_path=test_raster_path,
    output_path=masks_path,
    model_path=model_path,
    architecture="unet",
    encoder_name="resnet34",
    num_channels=3,
    num_classes=2,
    window_size=512,
    overlap=256,
    batch_size=4,
)
```

## Vectorize Masks

The model outputs a raster mask (each pixel = class ID: 0 or 1). For GIS workflows, we convert the binary mask to vector polygons using `orthogonalize`, which extracts connected regions and regularizes outlines.

```python
output_vector_path = str(data_dir / "naip_test_semantic_prediction.geojson")
gdf = geoai.orthogonalize(masks_path, output_vector_path, epsilon=2)
print(f"Extracted {len(gdf)} polygon(s) from semantic mask")
```

## Add Geometric Properties

Compute area, perimeter, and other properties for each polygon. Useful for filtering (e.g., remove tiny artifacts) and analysis.

```python
gdf_props = geoai.add_geometric_properties(gdf, area_unit="m2", length_unit="m")
print(gdf_props[["area_m2", "perimeter_m"]].head())
```

## Visualize and Filter Results

Visualize the predicted masks and vectors. Filtering by area (e.g., > 50 m²) removes small false positives.

```python
# Optional: Interactive visualization (uncomment if ipyleaflet is available)
geoai.view_raster(masks_path, nodata=0, basemap=test_raster_path, backend="ipyleaflet")
geoai.view_vector_interactive(gdf_props, column="area_m2", tiles=test_raster_path)
```

```python
gdf_filtered = gdf_props[gdf_props["area_m2"] > 50]
print(f"After filtering (area > 50 m²): {len(gdf_filtered)} polygon(s)")
```
